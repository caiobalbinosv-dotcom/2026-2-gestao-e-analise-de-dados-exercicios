## Importar Bibliotecas


```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

## Carregar Dados de entrada (X) e Saída (Y) - DADOS DE TREINO

```python
X_train = np.array([
    [2],
    [3],
    [4],
    [5],
    [6],
    [8],
    [10]
])

y_train = np.array([
    18,
    21,
    25,
    28,
    32,
    38,
    45
])
```

## Carregar Dados de entrada (X) e Saída (Y) - Dados de Teste

```python
X_test = np.array([
    [3.5],
    [7],
    [9]
])

y_test = np.array([
    23,
    35,
    42
])
```

## Escolher Tipo de Modelo

```python
modelo = LinearRegression()
```

## Fazer o Treino dos dados de treino (Gerar seu modelo)

```python
modelo.fit(X_train, y_train)

w = modelo.coef_[0]
b = modelo.intercept_
```

## Escrever os valores dos parâmetros "w" e "b" na tela

```python
print("w =", w)
print("b =", b)
```

```python


y_pred = modelo.predict(X_test)

print("Previsões:", y_pred)

mse = mean_squared_error(y_test, y_pred)

print("MSE teste =", mse)

nova_distancia = np.array([[7.5]])

tempo_estimado = modelo.predict(nova_distancia)

print("Tempo estimado:", tempo_estimado[0])
```
